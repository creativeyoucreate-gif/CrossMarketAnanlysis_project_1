# Top_3_CRYPTOCURRENCIES_Historical_prices

```python
import requests
import pandas as pd

coin_id = ["tether", "bitcoin", "ethereum"]
all_coins = []

for i in coin_id:
    coin_info_url = f"https://api.coingecko.com/api/v3/coins/{i}/market_chart?vs_currency=inr&days=365"
    response = requests.get(coin_info_url)
    coins = response.json()
    
    # Each entry in 'prices' is [timestamp, price]
    coin_df = pd.DataFrame(coins['prices'], columns=['Date', 'Price'])
    coin_df['Coin_id'] = i  # add coin name
    
    # Convert timestamp to date
    coin_df['Date'] = pd.to_datetime(coin_df['Date']).dt.date
    
    # Reorder columns: Coin_id, Date, Price
    coin_df = coin_df[['Coin_id', 'Date', 'Price']]
    all_coins.append(coin_df)
    

# Combine all coins into one DataFrame
final_df = pd.concat(all_coins, ignore_index=True)

# Sample 30 random rows
print(final_df.sample(30))
```

```python
import yfinance as yf
import pandas as pd

#choose indices
tickers = ['^GSPC', '^DJI', '^IXIC']  # S&P 500, Dow Jones, NASDAQ
start_date = '2020-01-01'
end_date = '2026-01-31'


#download historical data   
stocks_df=yf.download(tickers, start=start_date, end=end_date, group_by='tickers')
stocks_df.head()
```

```python
stocks_df1=stocks_df["^GSPC"].reset_index().dropna()

stocks_df1.head()
```

```python
stocks_df2=stocks_df["^DJI"].reset_index().dropna()
stocks_df2.head()
```

```python
stocks_df3=stocks_df["^IXIC"].reset_index().dropna()
stocks_df3.head()
```

```python
#Combine all stocks into one DataFrame
Stocks_prices_df = pd.concat([stocks_df1, stocks_df2, stocks_df3],
    keys=['^GSPC', '^DJI', '^IXIC'],
    names=['Ticker']).reset_index(level='Ticker').dropna()



# Reorder columns
Stocks_prices_df = Stocks_prices_df[['Date', 'Open', 'High', 'Low', 'Close', 'Volume', 'Ticker']]

Stocks_prices_df = pd.DataFrame(Stocks_prices_df.values,columns=['Date', 'Open', 'High', 'Low', 'Close', 'Volume', 'Ticker'])

print(Stocks_prices_df.sample(20))
```

```python
Stocks_prices_df.to_csv('stocks_price_history.csv', index=False)
```
